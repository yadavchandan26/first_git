```python
import pandas as pd
from sklearn.datasets import make_moons
```

```python
x,y=make_moons(
    n_samples=1000,
    noise=.01,
    random_state=42
)
```

```python
from sklearn.preprocessing import StandardScaler
x_scale=StandardScaler().fit_transform(x)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.scatterplot(x=x_scale[:,0],y=x_scale[:,1])
```

```python
from sklearn.cluster import DBSCAN
from sklearn.neighbors import NearestNeighbors
import numpy as np

min_samples=5
neigh=NearestNeighbors(n_neighbors=min_samples)
neigh.fit(x_scale)

distance,indices=neigh.kneighbors(x_scale)
k_distance=np.sort(distance[:,-1])
plt.plot(k_distance)
```

```python
model=DBSCAN(
    eps=.5,
    min_samples=5
)
pred=model.fit_predict(x_scale)

sns.scatterplot(x=x_scale[:,0],y=x_scale[:,1],c=pred)
```
